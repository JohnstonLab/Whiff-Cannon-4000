# Odour gun controller

```python
import pandas as pd
import matplotlib.pyplot as plt
import odour
```

## Load odour sequence protocol

```python
protocol_path = 'Protocols/Test.csv'
trials = pd.read_csv(protocol_path)
trials
```

## Set up vials and launch gui
- Reload if a new protocol is loaded

```python
odours = [f'Valve {i + 1}:' for i in range(12)]
odgun = odour.Gun(odours=odours, trials=trials)
odgun.gui()
```
